```python
def through_microscope(I0, a1, b1, b2, c1, c2, d1, d2, a1_, b1_, b2_, c1_, c2_, d1_, d2_):
    J1 = torch.stack([
        torch.stack([a1, b1 + 1j*b2]),
        torch.stack([c1 + 1j*c2, d1 + 1j*d2])
    ]
    ).to(torch.complex128)
    J2 = torch.stack([
        torch.stack([a1_, b1_ + 1j*b2_]),
        torch.stack([c1_ + 1j*c2_, d1_ + 1j*d2_])
    ]
    ).to(torch.complex128)
    phii = (torch.linspace(np.pi/2, 1.5*np.pi, 37).to(device))[:-1]+torch.tensor(-0*np.pi/180.).to(device)
    input_ = I0*torch.stack([torch.cos(phii), torch.sin(phii)]).to(device).to(torch.complex128)
    output_ = torch.einsum('ab,bc->ac', J1, input_)
    output_2 =  torch.einsum('ab,bc->ac', J2, input_)
    return (output_[0]*torch.conj(output_[0])).to(torch.float64), (output_[1]*torch.conj(output_[1])).to(torch.float64), (output_2[0]*torch.conj(output_2[0])).to(torch.float64), (output_2[1]*torch.conj(output_2[1])).to(torch.float64)
```

```python
res1 = torch.tensor(res1, device=device)
res2 = torch.tensor(res2, device=device)
res3 = torch.tensor(res3, device=device)
res4 = torch.tensor(res4, device=device)
res5 = torch.tensor(res5, device=device)
res6 = torch.tensor(res6, device=device)
```

```python
params = torch.tensor([I0, a1, b1, b2, c1, c2, d1, d2, a1_, b1_, b2_, c1_, c2_, d1_, d2_], device=device)
params.requires_grad=True
optimizer = torch.optim.Adam([params], lr=.05)
loss_ = []
```

```python
def loss_f(para):
    I0, a1, b1, b2, c1, c2, d1, d2, a1_, b1_, b2_, c1_, c2_, d1_, d2_ = para[0], para[1], para[2], para[3], para[4], para[5], para[6], para[7] , para[8], para[9], para[10], para[11], para[12] , para[13], para[14]
    th1, th2, th3, th4 = through_microscope(I0, a1, b1, b2, c1, c2, d1, d2, a1_, b1_, b2_, c1_, c2_, d1_, d2_) 
    J1 = torch.stack([
        torch.stack([a1, b1 + 1j*b2]),
        torch.stack([c1 + 1j*c2, d1 + 1j*d2])
    ]
    ).to(torch.complex128)   
    J2 = torch.stack([
        torch.stack([a1_, b1_ + 1j*b2_]),
        torch.stack([c1_ + 1j*c2_, d1_ + 1j*d2_])
    ]
    ).to(torch.complex128)   
    force_unitary1 = 0*torch.sum(torch.abs((torch.einsum('ab, bc-> ac', torch.transpose(torch.conj(J1), 0, 1), J1)-torch.ones(2))**2))
    force_unitary2 = 0*torch.sum(torch.abs((torch.einsum('ab, bc-> ac', torch.transpose(torch.conj(J2), 0, 1), J2)-torch.ones(2))**2))
    return torch.sum((th1-res1)**2)+torch.sum((th2-res2)**2) + torch.sum((th3-res3)**2)+torch.sum((th4-res4)**2) + (para[0]-torch.abs(para[0]))**2 + (para[1]-torch.abs(para[1]))**2 + force_unitary1 + force_unitary2 
```

```python
for i in range(500):
    optimizer.zero_grad()
    loss=loss_f(params)
    loss.backward()  # Backpropagation
    optimizer.step()
    loss_.append(loss.cpu().detach().numpy())
```
